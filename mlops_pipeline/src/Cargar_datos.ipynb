# 1.2 Validación de datos (Proceso de verificar que los datos sean correctos, completos y útiles antes de ser procesados o almacenados. )

```python
#rangos de las variables numéricas
print("\nRangos de las variables numéricas:")
for col in df.select_dtypes(include=['int64', 'float64']).columns:
    print(f"{col}: {df[col].min():,} - {df[col].max():,}")  
print("=" * 100)
```

### Edad con valores incoherentes a la realidad

## Valores Nulos
Se observa que seis (6) de las columnas presentan valores nulos para su posterior tratamiento.

```python
print("\nValores nulos en cada columna:")   
print(df.isnull().sum())  
print("=" * 100) 
```

## Valores duplicados 
Se observa que no hay duplicidad de valores

```python
# ver duplicados
print("\nRegistros duplicados:")
print(df.duplicated().sum())
print("=" * 100) 
```

```python
# ver valores unicos
print("\nValores únicos por columna:")
for col in df.columns:
    unique_vals = df[col].nunique()
    print(f"{col}: {unique_vals} valores únicos")
print("=" * 100)    
```
